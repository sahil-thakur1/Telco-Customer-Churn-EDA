```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/sample_data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.info()
```

```python
df.head()
```

```python

df["TotalCharges"] = df["TotalCharges"].replace(" ","0")
df["TotalCharges"] = df["TotalCharges"].astype("float")
```

```python
def con(value):
  if value==1:
    return "yes"
  else:
      return "no"

df['SeniorCitizen'] = df["SeniorCitizen"].apply(con)
```

```python
df.head(30)
```

```python
ax=  sns.countplot(x = df['Churn'],data = df)
ax.bar_label(ax.containers[0])
plt.title("Count Of Customer By Churn")
plt.show
```

```python
sns.countplot(x = 'gender',data = df, hue = "Churn")
plt.title("Churn By Gender")
plt.show()
```

```python
sns.countplot(x = 'SeniorCitizen',data = df, hue = "Churn")
plt.title("Churn By Senior Citizen")
plt.show()
```

```python
sns.histplot(x = 'tenure',data = df, hue = "Churn")
plt.show
```

```python
ax= sns.countplot(x = 'Contract',data = df, hue = "Churn")
ax.bar_label(ax.containers[0])
plt.title("Churn Of Customer By Contract")
plt.show()
```

```python
df.columns.values
```

```python
columns = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

plt.figure(figsize=(15, 12))

for i, col in enumerate(columns, 1):
    plt.subplot(3, 3, i)   # 3 rows, 3 columns
    sns.countplot(x=col, data=df , hue = "Churn")
    plt.title(col)
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
ax= sns.countplot(x = 'PaymentMethod',data = df, hue = "Churn")
ax.bar_label(ax.containers[0])
plt.title("Churn Of Customer By Payment Method")
plt.show()
```
